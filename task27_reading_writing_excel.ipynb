```python
import pandas as pd
data = {"Name": ["Amit", "Priya", "Rahul", "Sneha", "Arjun"], "Age": [22, 23, 21, 24, 22], "City": ["Mumbai", "Pune", "Delhi", "Hyderabad", "Chennai"], "Salary": [30000, 35000, 28000, 40000, 32000]}
df = pd.DataFrame(data)
df.to_excel("employees.xlsx", index=False)
print(df)
```

```python
df = pd.read_excel("employees.xlsx")
print(df)
```

```python
df_sheet = pd.read_excel("employees.xlsx", sheet_name="Sheet1")
print(df_sheet)
```

```python
selected_data = df[["Name", "Salary"]]
print(selected_data)
```

```python
high_salary = df[df["Salary"] > 30000]
print(high_salary)
```

```python
sorted_data = df.sort_values(by="Salary", ascending=False)
print(sorted_data)
```

```python
high_salary.to_excel("high_salary_employees.xlsx", index=False)
print("Filtered data saved successfully!")
```

```python
df_selected = pd.read_excel("employees.xlsx", usecols=["Name", "City"])
print(df_selected)
```

```python
new_employee = pd.DataFrame({"Name": ["Kiran"], "Age": [25], "City": ["Pune"], "Salary": [38000]})
updated_df = pd.concat([df, new_employee], ignore_index=True)
updated_df.to_excel("updated_employees.xlsx", index=False)
print(updated_df)
```
